```python

# ============================================================
# AD23532 - PRINCIPLES OF DATA SCIENCE
# EX5: INFORMATION EXTRACTION USING spaCy
# ============================================================

# -------------------- IMPORT LIBRARIES -----------------------

import pandas as pd
import spacy
from collections import Counter

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

print("spaCy loaded successfully!")


# ============================================================
# TASK 1 & 2: LOAD AND CLEAN DATASET
# ============================================================

# Load Amazon Fine Food Reviews dataset
df = pd.read_csv(r"C:\Users\HDCO422184\Downloads\archive (6)\Reviews.csv")

print("\nOriginal Dataset Shape:", df.shape)

# Select Text column
reviews = df[["Text"]].copy()

# Remove missing/null values
reviews = reviews.dropna()

# Remove duplicate reviews
reviews = reviews.drop_duplicates()

# Retain first 10,000 reviews
reviews = reviews.head(10000)

print("Number of reviews after cleaning:", len(reviews))

print("\nFirst 5 reviews:")
print(reviews.head())


# ============================================================
# TASK 3: TOKENS, LEMMAS AND POS TAGS FOR FIVE REVIEWS
# ============================================================

print("\n\n========== TASK 3: POS TAGGING ==========")

for i, text in enumerate(reviews["Text"].head(5)):

    doc = nlp(text)

    print("\n" + "=" * 70)
    print("REVIEW", i + 1)

    print("\nOriginal Review:")
    print(text)

    print("\nTokens:")
    print([token.text for token in doc])

    print("\nLemmas:")
    print([token.lemma_ for token in doc])

    print("\nPOS Tags:")
    print([(token.text, token.pos_) for token in doc])


# ============================================================
# TASK 4: NOUNS, VERBS, ADJECTIVES AND ADVERBS
# ============================================================

print("\n\n========== TASK 4: POS CATEGORIES ==========")

for i, text in enumerate(reviews["Text"].head(5)):

    doc = nlp(text)

    nouns = [token.text for token in doc if token.pos_ == "NOUN"]

    verbs = [token.text for token in doc if token.pos_ == "VERB"]

    adjectives = [token.text for token in doc if token.pos_ == "ADJ"]

    adverbs = [token.text for token in doc if token.pos_ == "ADV"]

    print("\n" + "=" * 70)
    print("REVIEW", i + 1)

    print("Nouns:", nouns)
    print("Verbs:", verbs)
    print("Adjectives:", adjectives)
    print("Adverbs:", adverbs)


# ============================================================
# TASK 5: FREQUENCY OF MAJOR POS CATEGORIES
# ============================================================

print("\n\n========== TASK 5: POS FREQUENCY ==========")

pos_counter = Counter()

for text in reviews["Text"]:

    doc = nlp(text)

    for token in doc:

        if token.pos_ in ["NOUN", "VERB", "ADJ", "ADV"]:

            pos_counter[token.pos_] += 1


pos_df = pd.DataFrame(
    pos_counter.items(),
    columns=["POS Category", "Frequency"]
)

pos_df = pos_df.sort_values(
    by="Frequency",
    ascending=False
)

print(pos_df)


# ============================================================
# TASK 6: NAMED ENTITY RECOGNITION (NER)
# ============================================================

print("\n\n========== TASK 6: NAMED ENTITY RECOGNITION ==========")

entities = []

for text in reviews["Text"]:

    doc = nlp(text)

    for ent in doc.ents:

        if ent.label_ in [
            "PERSON",
            "ORG",
            "DATE",
            "MONEY",
            "QUANTITY",
            "PRODUCT"
        ]:

            entities.append({
                "Review": text,
                "Entity": ent.text,
                "Label": ent.label_
            })


entity_df = pd.DataFrame(entities)

print("\nExtracted Entities:")
print(entity_df.head(20))


print("\nEntity Type Frequency:")

if not entity_df.empty:
    print(entity_df["Label"].value_counts())
else:
    print("No matching entities found.")


# ============================================================
# TASK 7: INFORMATION EXTRACTION FUNCTION
# ============================================================

print("\n\n========== TASK 7: INFORMATION EXTRACTION FUNCTION ==========")


def extract_information(review):

    doc = nlp(review)

    # Extract tokens
    tokens = [token.text for token in doc]

    # Extract lemmas
    lemmas = [token.lemma_ for token in doc]

    # Extract POS tags
    pos_tags = [
        (token.text, token.pos_)
        for token in doc
    ]

    # Extract named entities
    named_entities = [
        ent.text
        for ent in doc.ents
    ]

    # Extract entity labels
    entity_labels = [
        (ent.text, ent.label_)
        for ent in doc.ents
    ]

    return {
        "Tokens": tokens,
        "Lemmas": lemmas,
        "POS Tags": pos_tags,
        "Named Entities": named_entities,
        "Entity Labels": entity_labels
    }


print("Function created successfully!")


# ============================================================
# TASK 8: TEST FUNCTION USING THREE NEW REVIEWS
# ============================================================

print("\n\n========== TASK 8: TESTING FUNCTION ==========")

new_reviews = [

    "I bought a Samsung coffee maker for $100 yesterday.",

    "John ordered 2 kg of organic coffee from Amazon.",

    "The product was delivered on 15 September 2026 and it tastes great."

]


for i, review in enumerate(new_reviews):

    result = extract_information(review)

    print("\n" + "=" * 70)

    print("NEW REVIEW", i + 1)

    print("\nReview:")
    print(review)

    print("\nTokens:")
    print(result["Tokens"])

    print("\nLemmas:")
    print(result["Lemmas"])

    print("\nPOS Tags:")
    print(result["POS Tags"])

    print("\nNamed Entities:")
    print(result["Named Entities"])

    print("\nEntity Labels:")
    print(result["Entity Labels"])


# ============================================================
# TASK 9: DISPLAY INFORMATION IN STRUCTURED FORMAT
# ============================================================

print("\n\n========== TASK 9: STRUCTURED OUTPUT ==========")

results = []

for review in new_reviews:

    result = extract_information(review)

    results.append({

        "Review": review,

        "Tokens": result["Tokens"],

        "Lemmas": result["Lemmas"],

        "POS Tags": result["POS Tags"],

        "Named Entities": result["Named Entities"],

        "Entity Labels": result["Entity Labels"]

    })


result_df = pd.DataFrame(results)

print(result_df)


# ============================================================
# SAVE FINAL OUTPUT
# ============================================================

result_df.to_csv(
    "Information_Extraction_Output.csv",
    index=False
)

print("\n\nOutput saved as:")
print("Information_Extraction_Output.csv")

print("\n========== EX5 COMPLETED SUCCESSFULLY ==========")
```
